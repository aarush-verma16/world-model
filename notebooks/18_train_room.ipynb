| `configs/room_in_anchored.yaml` | yes | yes |

The 2026-11-17 run is archived under `checkpoints/room_2026-11-17` and `results/room_2026-11-17`. Run All starts fresh. Interrupt writes the checkpoint, and a later Run All resumes that new run.
